# UC06 Exploration
Exploration initiale de promotion.dat avec PySpark.

```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.master('local[2]').appName('UC06-exploration').getOrCreate()
path = '/opt/project/input/promotion/promotion.dat'
df = spark.read.option('sep', '|').option('header', 'false').csv(path)
print('Nombre de lignes :', df.count())
print('Nombre de colonnes :', len(df.columns))
df.show(3, truncate=False)
```

```python
df.printSchema()
print('Valeurs nulles field_00 :', df.filter(df['_c0'].isNull()).count())
print('Doublons field_00 :', df.count() - df.select('_c0').distinct().count())
```
